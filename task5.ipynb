```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/2016-weather-data-seattle.csv")
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
```

```python
sns.countplot(x='Month', data=df)
plt.title("Weather Records by Month")
plt.show()
```

```python
plt.plot(df['Date'], df['Max_TemperatureC'])
plt.title("Temperature Trend Over Time")
plt.show()
```

```python
sns.histplot(df['Max_TemperatureC'], bins=20)
plt.title("Temperature Distribution")
plt.show()
```

## 📊 Key Insights

- Weather patterns vary significantly across months  
- Temperature trends clearly show seasonal variations  
- Extreme temperature conditions may impact travel and safety  
- Data analysis helps identify patterns useful for planning and decision-making  
- Seasonal patterns can be useful for traffic and safety preparedness  

- Adverse weather conditions can lead to increased accident risks  
- Seasonal trends help in planning traffic safety measures
